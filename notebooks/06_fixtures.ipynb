```python
import sys; sys.path.insert(0, "../src")
import pandas as pd
from fixtures import load_pending

df = pd.read_csv("../data/processed/laliga_matches.csv",
                 parse_dates=["Date"], dtype={"Season": str})
played = df[df["Season"] == "2627"]

fx = load_pending("../data/raw/fixtures_2627.csv", played)
print("Partidos en el calendario:", len(fx))
print("Jugados:", (~fx["pendiente"]).sum(), "| Pendientes:", fx["pendiente"].sum())

betis = fx[(fx["HomeTeam"] == "Betis") | (fx["AwayTeam"] == "Betis")]
print("Betis pendientes:", betis["pendiente"].sum())
fx[fx["pendiente"]].head()
```
